# Publish the Virtual Prof to Hugging Face

This notebook copies the two small language models used by the **Virtual Prof** (backup chapter A of *Materials & the Environment*) and the book's knowledge base to the `MaterialsandtheEnvironment` organisation on Hugging Face.

1. Create a **write** token at <https://huggingface.co/settings/tokens> (with access to the organisation).
2. Run the cells. Nothing is trained: the models are copied as they are (Apache-2.0), with a model card.

The page works before this is done (it falls back to the public `onnx-community` builds); after it, everything is served from your organisation.

```python
!pip -q install -U huggingface_hub
```

```python
import os, urllib.request
from getpass import getpass
from huggingface_hub import HfApi
token = os.environ.get("HF_TOKEN") or getpass("Hugging Face write token: ")
api = HfApi(token=token)
print("Logged in as", api.whoami()["name"])
```

```python
# get the publishing script and the knowledge base from the book's repository
BASE = "https://raw.githubusercontent.com/MaterialsandtheEnvironment/MaterialsandtheEnvironment.github.io/main/"
os.makedirs("tools", exist_ok=True); os.makedirs("data", exist_ok=True)
urllib.request.urlretrieve(BASE + "tools/publish_virtual_prof.py", "tools/publish_virtual_prof.py")
urllib.request.urlretrieve(BASE + "data/prof_index.json", "data/prof_index.json")
print(open("tools/publish_virtual_prof.py").read()[:600])
```

```python
!python3 tools/publish_virtual_prof.py --token $token
```
